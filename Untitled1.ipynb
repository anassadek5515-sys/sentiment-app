```python
# 1. تثبيت المكتبات المطلوبة للتشغيل السريع
!pip install -q transformers accelerate torch bitsandbytes

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline

# 2. تحميل نموذج مفتوح المصدر قوي وسريع جداً للتجارب
model_id = "Qwen/Qwen1.5-0.5B-Chat"

print("⏳ جاري تحميل النموذج على كارت الشاشة (GPU)...")
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(
    model_id,
        torch_dtype=torch.float16,
            device_map="auto"
            )

            # 3. إعداد أنبوب المحادثة وتجربة أول سؤال
pipe = pipeline("text-generation", model=model, tokenizer=tokenizer)

prompt = "ما هي أول خطوة لبناء نموذج ذكاء اصطناعي منافس؟"
messages = [
                {"role": "system", "content": "أنت خبير ومساعد ذكاء اصطناعي احترافي."},
                    {"role": "user", "content": prompt}
                    ]
formatted_prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
outputs = pipe(formatted_prompt, max_new_tokens=100, do_sample=True, temperature=0.7)

print("\n🤖 إجابة النموذج:")
print(outputs[0]["generated_text"])
```

```python
# كود تحويل النموذج إلى Agent يحكم ويفكر قبل الإجابة
def run_ai_agent(task_description):
    print("🤖 [Agent] جاري تحليل المهارات والمهمة المطلوبة...")

            # صياغة التعليمات للعميل الذكي
agent_system_prompt = (
                        "أنت عميل ذكاء اصطناعي مستقل (AI Agent) متصل بأدوات تطوير. "
                                "مهمتك هي تفكيك أي مشكلة تقنية إلى 3 خطوات تنفيذية محددة وعملية جداً. "
                                        "أجب باللغة العربية فقط."
                                            )

messages = [
                                                            {"role": "system", "content": agent_system_prompt},
                                                                    {"role": "user", "content": f"المهمة المطلوبة: {task_description}"}
                                                                        ]

formatted = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)

out = pipe(
                                                                                                formatted,
                                                                                                        max_new_tokens=250,
                                                                                                                do_sample=True,
                                                                                                                        temperature=0.3,
                                                                                                                                repetition_penalty=1.2
                                                                                                                                    )

full_text = out[0]["generated_text"]
response = full_text.split("<|im_start|>assistant\n")[-1].replace("<|im_end|>", "")
return response

                                                                                                                                                    # تجربة تكليف الـ Agent بمهمة معقدة
task = "تصميم نظام ذكاء اصطناعي لتحليل بيانات العملاء وتوقع سلوك الشراء."
print("📋 خطة عمل الـ Agent:\n")
print(run_ai_agent(task))
```
